# GestureFusion — Demo
This notebook demonstrates loading synthetic sensor data and plotting basic features.

```python
import matplotlib.pyplot as plt
import numpy as np
t = np.linspace(0, 2*np.pi, 100)
x = np.sin(t)
y = np.cos(t)
plt.plot(t,x,label='acc_x')
plt.plot(t,y,label='acc_y')
plt.legend()
plt.show()
```
